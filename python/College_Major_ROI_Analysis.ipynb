# College Major ROI — Senior Data Analyst Deep Dive
This notebook performs EDA, KPI analysis, segmentation, relationship analysis, and business recommendations using Python. SQL equivalents are in `sql/business_queries.sql`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('college_major_roi.csv')
df.head()
```

## 1. Initial EDA

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all').T)
display(df.isna().sum().sort_values(ascending=False))
print('Duplicate grad_id:', df['grad_id'].duplicated().sum())
```

## 2. Business KPIs

```python
kpis = pd.Series({
    'Average Net ROI': df.net_roi_usd.mean(),
    'Median Net ROI': df.net_roi_usd.median(),
    'Positive ROI Rate': df.positive_roi.mean(),
    'High ROI Rate': df.high_roi.mean(),
    'Average Net Cost': df.net_cost_usd.mean(),
    'Average Debt': df.debt_usd.mean(),
    'Average Added Earnings': df.added_earnings_10yr_usd.mean()
})
kpis
```

## 3. ROI by Major

```python
major_perf = df.groupby('major').agg(graduates=('grad_id','size'), avg_net_roi=('net_roi_usd','mean'), median_net_roi=('net_roi_usd','median'), positive_roi_rate=('positive_roi','mean'), high_roi_rate=('high_roi','mean')).sort_values('avg_net_roi', ascending=False)
major_perf
```

## 4. Internship, completion, institution tier and selectivity

```python
display(df.groupby('had_internship').agg(n=('grad_id','size'), avg_roi=('net_roi_usd','mean'), positive=('positive_roi','mean')))
display(df.groupby('completed_on_time').agg(n=('grad_id','size'), avg_roi=('net_roi_usd','mean'), positive=('positive_roi','mean')))
display(df.groupby('institution_tier').agg(n=('grad_id','size'), avg_roi=('net_roi_usd','mean'), positive=('positive_roi','mean')).sort_values('avg_roi', ascending=False))
```

## 5. Correlation and interpretation

```python
df.select_dtypes('number').corr()['net_roi_usd'].sort_values(ascending=False)
```

## 6. Visual analysis

```python
fig, ax = plt.subplots(figsize=(10,6))
ax.hist(df.net_roi_usd, bins=50)
ax.axvline(0, linestyle='--')
ax.set_title('Distribution of 10-Year Net ROI')
ax.set_xlabel('Net ROI (USD)')
ax.set_ylabel('Graduates')
plt.tight_layout()
plt.show()
```
